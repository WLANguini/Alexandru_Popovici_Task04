```python
lista_coloane = [df["drive_unit"], df["transmission"], df["fuel_type"]]

for i in lista_coloane:
    plt.plot(size = [12, 24])
    sns.boxplot(x = i, y= df["priceUSD"])
    plt.xticks(rotation = 45)
    plt.title("Price Correlation")
    plt.xlabel("Corr factors")
    plt.ylabel("Price")
```
